```python
total_revenue = df["TotalPrice"].sum()
total_orders = df["InvoiceNo"].nunique()
average_order_value = total_revenue / total_orders
average_order_value
```

```python
customer_orders = df.groupby("CustomerID")["InvoiceNo"].nunique().sort_values(ascending=False)
customer_orders.head(10)
```

```python
customer_orders = df.groupby("CustomerID")["InvoiceNo"].nunique()
repeat_customers = customer_orders[customer_orders > 1]
repeat_customers
```

```python
repeat_customers.count()
```

```python
one_time_customers = customer_orders[customer_orders == 1]
one_time_customers.count()
```

```python
customer_spending = df.groupby("CustomerID")["TotalPrice"].sum().sort_values(ascending=False)
customer_spending.head(10)
```

```python
customer_spending = df.groupby("CustomerID")["TotalPrice"].sum()
high_value = customer_spending[customer_spending >= 5000]
medium_value = customer_spending[
    (customer_spending >= 1000) &
    (customer_spending < 5000)
]
low_value = customer_spending[customer_spending < 1000]
```

```python
print("High Value:", len(high_value))
print("Medium Value:", len(medium_value))
print("Low Value:", len(low_value))
```

```python
country_analysis = df.groupby("Country").agg(
    Revenue=("TotalPrice", "sum"),
    Orders=("InvoiceNo", "nunique")
)
country_analysis["Average_Order_Value"] = (
    country_analysis["Revenue"] / country_analysis["Orders"]
)
country_analysis = country_analysis.sort_values(
    "Revenue", ascending=False
)
country_analysis
```

```python
product_analysis = df.groupby("Description").agg(
    Revenue=("TotalPrice", "sum"),
    Orders=("InvoiceNo", "nunique")
)
product_analysis = product_analysis.sort_values(
    "Revenue", ascending=False
)
product_analysis.head(10)
```

```python
High_value_transactions = df.sort_values("TotalPrice", ascending=False).head(10)
High_value_transactions
```

```python
df["Year"] = pd.to_datetime(df["Invoice Date"]).dt.year
yearly_revenue = df.groupby("Year")["TotalPrice"].sum()
yearly_revenue
```

```python
df["Hour"] = pd.to_datetime(df["Invoice Time"].astype(str)).dt.hour
hourly_sales = df.groupby("Hour")["TotalPrice"].sum().sort_values(ascending=False)
hourly_sales
```

```python
text_columns = [ "StockCode", "Description", "Country",]

df[text_columns] = df[text_columns].apply(lambda x: x.str.title())
```

```python
total_revenue = df["TotalPrice"].sum()

total_orders = df["InvoiceNo"].nunique()

total_customers = df["CustomerID"].nunique()

total_products = df["Description"].nunique()

total_quantity = df["Quantity"].sum()

average_order_value = total_revenue / total_orders
```

```python
summary_dashboard = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Orders",
        "Total Customers",
        "Total Products",
        "Total Quantity Sold",
        "Average Order Value"
    ],
    "Value": [
        total_revenue,
        total_orders,
        total_customers,
        total_products,
        total_quantity,
        average_order_value
    ]
})
```

```python
summary_dashboard
```

```python
#Monthly Revenue

df["Month"] = pd.to_datetime(df["Invoice Date"]).dt.to_period("M")
monthly_sales = df.groupby("Month")["TotalPrice"].sum().sort_values(ascending=False)
```

```python
def check_dtype(var, label):
    print(f"\n=== {label} ===")
    print("Type:", type(var))
    if isinstance(var, pd.Series):
        print("dtype:", var.dtype)
    elif isinstance(var, pd.DataFrame):
        print(var.dtypes)
    else:
        print("Not a Series or DataFrame")

check_dtype(summary_dashboard, "summary_dashboard")
check_dtype(df, "df")
check_dtype(Top10Customers, "Top10Customers")
check_dtype(total_revenue, "total_revenue")
check_dtype(total_orders, "total_orders")
check_dtype(top_products_revenue, "top_products_revenue")
check_dtype(top_products_quantity, "top_products_quantity")
check_dtype(Transactionspercountry, "Transactionspercountry")
check_dtype(High_value_transactions, "High_value_transactions")
check_dtype(country_revenue, "country_revenue")
check_dtype(average_order_value, "average_order_value")
check_dtype(country_analysis, "country_analysis")
check_dtype(product_analysis, "product_analysis")
check_dtype(customer_orders, "customer_orders")
check_dtype(customer_spending, "customer_spending")
check_dtype(repeat_customers, "repeat_customers")
check_dtype(one_time_customers, "one_time_customers")
check_dtype(hourly_sales, "hourly_sales")
check_dtype(monthly_sales, "monthly_sales")
check_dtype(yearly_revenue, "yearly_revenue")
check_dtype(cancellations_per_country, "cancellations_per_country")
check_dtype(top10_cancellation_countries, "top10_cancellation_countries")
```

```python
with pd.ExcelWriter("Retail_Analysis_Report.xlsx", engine="openpyxl") as writer:

    summary_dashboard.to_excel(writer, sheet_name="Summary Dashboard", index=False)
    df.to_excel(writer, sheet_name="Cleaned Data", index=False)
    Top10Customers.to_excel(writer, sheet_name="Top 10 Customers")
    total_revenue.to_excel(writer, sheet_name="Total Revenue")
    total_orders.to_excel(writer, sheet_name="Total Orders")
    top_products_revenue.to_excel(writer, sheet_name="Top Products Revenue")
    top_products_quantity.to_excel(writer, sheet_name="Top Products Quantity")
    Transactionspercountry.to_excel(writer, sheet_name= "Transactions per country")
    High_value_transactions.to_excel(writer, sheet_name="High Value Transactions")
    country_revenue.to_excel(writer, sheet_name= "Revenue per Country")
    average_order_value.to_excel(writer, sheet_name="Average Order Value")
    country_analysis.to_excel(writer, sheet_name="Country Analysis")
    product_analysis.to_excel(writer, sheet_name="Product Analysis")
    customer_orders.to_excel(writer, sheet_name="Customer Orders")
    customer_spending.to_excel(writer, sheet_name="Customer Spending")
    repeat_customers.to_excel(writer, sheet_name="Repeat Customers")
    one_time_customers.to_excel(writer, sheet_name="One Time Customers")
    hourly_sales.to_excel(writer, sheet_name="Hourly Sales")
    monthly_sales.to_excel(writer, sheet_name="Monthly Sales")
    yearly_revenue.to_excel(writer, sheet_name="Annual Revenue")
    cancellations_per_country.to_excel(writer, sheet_name="Cancellation per Country")
    top10_cancellation_countries.to_excel(writer, sheet_name="Top 10 Cancellations")


print("Excel report created successfully!")
```
